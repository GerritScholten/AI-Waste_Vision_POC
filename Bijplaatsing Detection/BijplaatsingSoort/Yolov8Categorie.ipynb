```python
# === SETUP ===
from ultralytics import YOLO
import os
import random
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix
from PIL import Image
```

```python
dataset_path = os.getcwd()  # huidige map
print("Current folder:", dataset_path)

print("\nFiles in folder:")
print(os.listdir(dataset_path))

print("\nImages count:", len(os.listdir("images")))
print("Labels count:", len(os.listdir("labels")))

# Check of elke image een bijbehorende label heeft
img_files = {f.split('.')[0] for f in os.listdir("images")}
lbl_files = {f.split('.')[0] for f in os.listdir("labels")}

missing_labels = img_files - lbl_files

print("\nMissing labels:", missing_labels if missing_labels else "Geen missende labels!")
```

```python
model = YOLO("yolov8n.pt")  # klein & snel model voor start

results = model.train(
    data="data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    name="categorie_detection",
)

print("Training klaar!")
```

```python
model_eval = YOLO("runs/detect/categorie_detection/weights/best.pt")
metrics = model_eval.val()

print("\n=== EVALUATION METRICS ===")
print(metrics)
```

```python
sample_imgs = random.sample(os.listdir("images"), 10)

for img_name in sample_imgs:
    img_path = f"images/{img_name}"
    print("\nAfbeelding:", img_path)
    model_eval(img_path, show=True)
```
