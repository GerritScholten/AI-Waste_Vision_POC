```python
import os
import shutil
import random
from pathlib import Path

base = Path("C:/Users/rtvan/Documents/GitHub/AI-Waste_Vision_POC/Bijplaatsing Detection/BijplaatsingSoort/MergedData_v2")
img_dir = base / "images"
lbl_dir = base / "labels"

train_dir = base / "train"
val_dir = base / "val"
test_dir = base / "test"

for d in [train_dir, val_dir, test_dir]:
    (d / "images").mkdir(parents=True, exist_ok=True)
    (d / "labels").mkdir(parents=True, exist_ok=True)

# Parameters
val_split = 0.2
test_split = 0.1

# Alle afbeeldingen ophalen
images = [f for f in os.listdir(img_dir) if f.lower().endswith((".jpg", ".png", ".jpeg"))]
random.shuffle(images)

n_total = len(images)
n_test = int(n_total * test_split)
n_val = int(n_total * val_split)
n_train = n_total - n_test - n_val

test_imgs = images[:n_test]
val_imgs = images[n_test:n_test+n_val]
train_imgs = images[n_test+n_val:]

def copy_files(file_list, target):
    for img in file_list:
        label = img.rsplit('.', 1)[0] + ".txt"
        shutil.copy(img_dir / img, target / "images" / img)
        shutil.copy(lbl_dir / label, target / "labels" / label)

copy_files(train_imgs, train_dir)
copy_files(val_imgs, val_dir)
copy_files(test_imgs, test_dir)

print("Dataset splits succesvol aangemaakt:")
print(f"Train: {len(train_imgs)} images")
print(f"Val:   {len(val_imgs)} images")
print(f"Test:  {len(test_imgs)} images")
```

```python
from ultralytics import YOLO
from pathlib import Path

model = YOLO("runs/obb/train5/weights/best.pt")
folder = Path("testimages")

print("Voorspellingen:\n")

for img in folder.glob("*.*"):
    results = model(img, imgsz=640, conf=0.20)[0]

    names = model.names

    # OBB predictions zitten in results.obb, NIET results.boxes
    obb = results.obb

    if obb is not None and obb.cls is not None:
        class_ids = obb.cls.int().tolist()
        unique = sorted(set(class_ids))
        categories = [names[i] for i in unique]
    else:
        categories = []

    print(f"{img.name}: {categories}")
```

```python
import os
from pathlib import Path

# pad naar je volledige dataset (labels)
label_dir = Path("MergedData_v2/labels")

# jouw klassen uit data.yaml
class_names = {
    0: "Container",
    1: "GFT",
    2: "Glas",
    3: "Papier",
    4: "Plastic",
    5: "Rest",
    6: "Voertuig"
}

# teller
counts = {name: 0 for name in class_names.values()}

# labelbestanden uitlezen
for label_file in label_dir.glob("*.txt"):
    with open(label_file, "r") as f:
        for line in f.readlines():
            parts = line.strip().split()
            if len(parts) > 0:
                cls_id = int(parts[0])
                if cls_id in class_names:
                    counts[class_names[cls_id]] += 1

# resultaat printen
print("Aantal objecten per categorie in volledige dataset:\n")
for cls, count in counts.items():
    print(f"{cls}: {count}")
```
